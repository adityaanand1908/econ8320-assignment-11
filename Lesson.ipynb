```python
"""
This file represents the 3D scatter plot stating Number of Races wins by those drivers,
 who has low crash rates
"""

import plotly.express as px
import pandas as pd

data = pd.read_csv("https://raw.githubusercontent.com/adityaanand1908/econ8320-assignment-11/main/nascar_racing_data_extended.csv")

plot = px.scatter(data, x='Crash Rate (%)', y='Wins',
           title="Lower Crash Rate wins the NASCAR Race",
           labels={
               'Crash Rate (%)' : 'Number of Crashes',
               'Wins' : 'Number of Wins'
           },
            trendline='ols'
           )

plot.show()

```

```python
"""
This file represents the 2D bar plot stating which type of cars irrespective of track type help
drivers win the race
"""

import plotly.express as px
import pandas as pd

data = pd.read_csv("https://raw.githubusercontent.com/adityaanand1908/econ8320-assignment-11/main/nascar_racing_data_extended.csv")

plot = px.bar(data,
                 x='Car Type',
                 y='Wins',
                 title='Cars that Wins On All Track',
                 labels={
                     'Car Type': 'Type of Cars',
                     'Wins': 'Number of Wins'
                 })


plot.show()
```

```python
"""
This file represents the 3D scatter plot stating Number of Races wins by those drivers, who has low crash rates
and high pass efficiency
"""

import plotly.express as px
import pandas as pd

data = pd.read_csv("https://raw.githubusercontent.com/adityaanand1908/econ8320-assignment-11/main/nascar_racing_data_extended.csv")

plot = px.scatter_3d(data,
                    x='Crash Rate (%)',
                    y='Pass Efficiency (NDOs)',
                    z='Wins',
                    color='Wins',
                    title='Wins vs. Crash Rate and Pass Efficiency',
                    labels={
                        'Crash Rate (%)': 'Number of Crashes in RaceCrash Rate',
                        'Pass Efficiency (NDOs)': 'Pass Efficiency',
                        'Wins': 'Number of Wins ↑'
                    })
plot.show()
```
